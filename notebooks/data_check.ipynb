# Data check: reading SAS transport (.xpt) files

`pandas.read_sas` reads `.xpt` files directly (needs only pandas).

```python
import pandas as pd

df = pd.read_sas("BPX_H.xpt", format="xport", encoding="utf-8")
print(df.shape)
df.head()
```

```python
df.info()
df.describe().T
```

Alternative: `pyreadstat` also returns variable labels (NHANES files usually include them).  
Install with `pip install pyreadstat`.

```python
df_2 = pd.read_sas("DEMO_H.xpt", format="xport", encoding="utf-8")
print(df_2.shape)
df_2.head()
```

```python
df_2.info()
df_2.describe().T
```

```python
joined_df = pd.merge(df, df_2, on="SEQN", how="inner")  
print(joined_df.shape)
joined_df.head()
```

```python
joined_df.describe().T
```
